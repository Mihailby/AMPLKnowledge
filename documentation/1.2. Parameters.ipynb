### 1.1. Parameters
Структура, Имя, definition or declaring

```python
param {indexing-opt} paramname {alias-opt} {attributes-opt} ;
```

Where:
- `indexing-opt `: optional indexing expression of the form `{indexname in indexset}` or `{indexset}`
- `paramname`: name of the parameter being declared
- `alias-opt`: optional alternate name, introduced by `alias`
- `attributes-opt`: zero or more attribute specifications (e.g., `>=`, `<=`, `default`, `integer`, `symbolic`, `in`)

Examples:

```python
param cost;
param price{PROD} >= 0;
param max_iteration integer default 100;
param filename symbolic;
param lower_bound {PROD} >= 0 <= 100;
```
