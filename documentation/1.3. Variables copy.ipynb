### 1.4. Constraints
Структура, Имя, definition or declaring

```python
subject to {indexing-opt} constraintname {alias-opt} expression {= | <= | >=} expression ;
```

Where:
- `indexing-opt`: optional indexing expression of the form `{indexname in indexset}` or `{indexset}`
- `constraintname`: name of the constraint being declared
- `alias-opt`: optional alternate name, introduced by alias
- `attributes-opt`: zero or more attribute specifications (e.g., `>=`, `<=`, `=` for fixed variables, `integer`, `binary`, `default` initialization)

Examples:

```python
var x;
var Sell{PROD} >= 0;
var y integer >= 1 <= 10;
var flag binary;
var fixed_var = 42;               # Fixed variable (:= syntax also allowed)
var init_var := 5.5;              # With initial guess
```
