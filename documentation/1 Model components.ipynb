### 1. Model Components

AMPL (A Mathematical Programming Language) is a declarative algebraic modeling language. A model describes a decision problem in terms of five kinds of named components: **sets**, **parameters**, **variables**, **constraints**, and **objectives**. Each component is introduced by a declaration statement in the model file.

Sets and parameters together constitute the **data layer** — they represent known, fixed information. Variables are the **decision layer** — their values are determined by the solver. Constraints and objectives together form the **problem layer** — they define what constitutes a feasible and optimal solution.

AMPL separates the **model** (structural description) from the **data** (specific instance values). The same model file can be solved against different data files without modification.

#### Summary of model components

| Keyword | Component | Role |
|---|---|---|
| `set` | Set | A finite collection of distinct members used to index other components |
| `param` | Parameter | A fixed numeric or symbolic datum known before solving |
| `var` | Variable | An unknown quantity whose value is determined by the solver |
| `subject to` | Constraint | An equality or inequality relation the variables must satisfy |
| `minimize` / `maximize` | Objective | A scalar expression to be minimized or maximized (optional) |

---

#### Sets

A **set** is a finite collection of distinct members. Members may be strings, numbers, or tuples. Sets are the foundation of AMPL indexing: almost all parameters, variables, and constraints in a typical model are indexed over sets, and summations range over sets. A set declared without an explicit membership definition receives its data from a data file or programmatic assignment.

```python
set PROD;                        # declared; membership supplied in data
set SHIFTS = {"Day", "Eve", "Night"};  # defined inline
set LINKS within {PROD, PROD};   # subset constraint
```

#### Parameters

A **parameter** is a constant datum — numeric or symbolic — whose value is fixed before the optimization problem is solved. Parameters represent problem data such as costs, capacities, and bounds. A parameter indexed over a set defines a function from set members to values. Attribute expressions such as `>= 0` declare constraints on admissible values that AMPL enforces when data is loaded.

```python
param T integer > 0;             # scalar integer parameter
param price {PROD} >= 0;         # indexed parameter with lower bound
param rate default 1.0;          # scalar with default value
```

Learn more: [1.2. Parameters](<1.2. Parameters.ipynb>)

---

#### Variables

A **variable** represents a decision quantity whose value is unknown and is to be determined by the solver. Variables may be continuous, integer, or binary, and may carry lower and upper bounds. Syntactically, variable declarations follow the same structure as parameter declarations, but their values are set by the solver rather than by data.

```python
var Sell {PROD} >= 0;            # nonneg continuous variable indexed over PROD
var Use {PROD} >= 0 <= 500;      # bounded continuous variable
var Open {PROD} binary;          # binary (0/1) variable
```

Learn more: [1.3. Variables](<1.3. Variables.ipynb>)

#### Constraints

A **constraint** is an equality or inequality relation that feasible solutions must satisfy. Constraints define the feasible region of the problem. Each constraint declaration introduces a named relation between an expression in the variables and a bound or another expression. An indexed constraint declaration generates one constraint per member of the indexing set. The keyword `subject to` (abbreviated `s.t.`) is optional — AMPL recognizes constraint declarations by their structure.

```python
subject to Capacity:                        # scalar constraint
    sum {p in PROD} Sell[p] <= total_cap;

subject to MarketLimit {p in PROD}:         # indexed constraint
    Sell[p] <= market[p];
```

---

#### Objectives

An **objective** is a scalar expression in the variables that the solver is directed to minimize or maximize. A model may contain zero or more objective declarations. If no objective is present, the problem is a **feasibility problem** — the solver searches for any point satisfying the constraints. If multiple objectives are declared, AMPL uses the first by default; an alternative objective can be selected explicitly. The objective name serves as a reference to its current value after solving.

```python
maximize TotalProfit:
    sum {p in PROD} (price[p] - cost[p]) * Sell[p];

minimize TotalCost:
    sum {p in PROD} cost[p] * Sell[p];
```

<div class="alert alert-block alert-success"> <b>Best practice: </b><br>
Separate model structure from data. Declare all sets and parameters in the model file without inline data; supply values via <code>amplpy</code> or in <code>.dat</code> file. This keeps the model reusable across problem instances.</div>
