## Strict Comparisons Use a Tolerance

A strict comparison has no closed boundary, so MP replaces it with a non-strict one shifted by a
small tolerance, the option `cvt:mip:eps` (default `1e-4`). Buying extra capacity costs 20, so the model
keeps `x < 3` true and pushes `x` as close to 3 as the tolerance allows:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
var x >= 0, <= 10;                # continuous amount
var Extra binary;                 # 1 if the extra capacity is bought
subject to Rule: x < 3 or Extra = 1;
maximize Gain: x - 20 * Extra;
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# run.py
ampl.solve(solver="highs")
print("x =", ampl.var["x"].value())
```
