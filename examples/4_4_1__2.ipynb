## Inheriting Reverse Order

Suppose `PRIORITY` is an ordered master list. A subset can keep the opposite order. The data
gives `REVIEW` its members in any order; `to_list()` returns them in the order the declaration
gives the set:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PRIORITY ordered;                     # items from highest to lowest priority
set REVIEW ordered by reversed PRIORITY;  # same items, lowest priority first
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PRIORITY"] = ["urgent", "high", "normal", "low"]
ampl.set["REVIEW"] = ["urgent", "normal", "low"]
```

```python
# run.py
print(ampl.set["REVIEW"].to_list())
print(ampl.get_value("first(REVIEW)"))
```

### Pure AMPL: data.dat and model.run

Pure AMPL:

```python
%%writefile data.dat
set PRIORITY := urgent high normal low;
set REVIEW := urgent normal low;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display REVIEW, first(REVIEW);
```
