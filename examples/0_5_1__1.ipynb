## Start from the Colab Template

The [AMPL Model Colaboratory Template](https://colab.ampl.com/notebooks/ampl-model-colaboratory-template.html)
is the layout every AMPL Colab notebook follows: an install cell, the `ampl_notebook` cell, the
model in a `%%ampl_eval` cell, data from Python, the solve, and the results as a pandas
DataFrame. A `%%ampl_eval` cell is equivalent to `ampl.eval("""cell content""")`.

The template loads parameters from plain Python lists. A list assigned to a parameter indexed
over one set fills it in the order of that set's members:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set FOOD;                                  # foods that can be bought
param cost {FOOD} > 0;                     # cost per package
param f_max {FOOD} >= 0;                   # most packages of each food
var Buy {j in FOOD} >= 0, <= f_max[j];     # packages bought
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["FOOD"] = ["BEEF", "CHK", "FISH"]
ampl.param["cost"] = [3.59, 2.59, 2.29]    # one value per member, in set order
ampl.param["f_max"] = [10, 10, 10]
```

```python
# run.py
print(ampl.param["cost"].to_dict())
try:
    ampl.param["f_max"] = [10, 10]         # one value short
except TypeError as error:
    print(type(error).__name__, error)
```
