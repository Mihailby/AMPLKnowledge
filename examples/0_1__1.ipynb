## One Model, Different Data

The model below adds a market limit to the production decision. Python supplies one
instance's data, and `ampl.solve(solver="highs")` asks the HiGHS solver to optimize it.
`ampl.var["Make"].to_pandas()` then returns the chosen production quantities.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD;                                 # Products the mill can make
param profit {PROD};                      # Profit per ton
param market {PROD} >= 0;                 # Maximum sales in tons

var Make {p in PROD} >= 0, <= market[p];  # Tons to make, capped by sales

maximize TotalProfit:                     # Total profit in dollars
    sum {p in PROD} profit[p] * Make[p];
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PROD"] = ["bands", "coils"]
ampl.param["profit"] = {"bands": 25, "coils": 30}
ampl.param["market"] = {"bands": 6000, "coils": 4000}
```

```python
# run.py
ampl.solve(solver="highs")
print(ampl.solve_result)
print(ampl.var["Make"].to_pandas())
```

### Pure AMPL: data.dat and model.run

The same model can be run in pure AMPL. The data file supplies the instance, while the run file
selects HiGHS, solves, and displays the status and decisions:

```python
%%writefile data.dat
set PROD := bands coils;
param profit := bands 25  coils 30;
param market := bands 6000  coils 4000;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
option solver highs;
solve;
display solve_result;
display Make;
```
