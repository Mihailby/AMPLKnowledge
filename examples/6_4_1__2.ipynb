## Selecting Among Objectives

With several declared objectives, `objective Emissions;` is equivalent to dropping all
objectives and then restoring `Emissions`. Selecting the objective explicitly is safer than
relying on whichever one happens to be active. amplpy has no method for the `objective` command,
so the Python tab sends it with `ampl.eval()`:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD;
param demand {PROD} >= 0;
param cost {PROD} >= 0;
param co2 {PROD} >= 0;
var Make {p in PROD} >= 0;
minimize Cost: sum {p in PROD} cost[p] * Make[p];
minimize Emissions: sum {p in PROD} co2[p] * Make[p];
subject to Service {p in PROD}: Make[p] >= 0.8 * demand[p];
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PROD"] = ["bands", "coils", "plate"]
ampl.param["demand"] = {"bands": 10, "coils": 20, "plate": 5}
ampl.param["cost"] = {"bands": 2, "coils": 3, "plate": 4}
ampl.param["co2"] = {"bands": 3, "coils": 1, "plate": 2}
```

```python
# run.py
ampl.option["solver_msg"] = 0
for name in ["Cost", "Emissions"]:
    ampl.eval(f"objective {name};")
    ampl.solve(solver="highs")
    print(name, ampl.obj[name].value())
```

### Pure AMPL: data.dat and model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%writefile data.dat
set PROD := bands coils plate;
param: demand cost co2 :=
  bands 10 2 3
  coils 20 3 1
  plate  5 4 2;
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
option solver highs;
option solver_msg 0;
objective Cost;
solve;
display Cost;
objective Emissions;
solve;
display Emissions;
```
