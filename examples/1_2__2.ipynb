## Parameters over Pairs and Longer Tuples

A parameter indexed over two sets is like a table, with one value for every combination.
`param cost {ORIG, DEST};` has a value `cost[i,j]` for every origin `i` and destination `j`.

When only some combinations make sense, index the parameter over a set of pairs instead.
Then values exist only for the pairs in that set, like a sparse matrix. From Python, a set of
pairs is a list of tuples, and the parameter's dictionary is keyed by the same tuples:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set ORIG;
set DEST;
set LINKS within {ORIG, DEST};   # only the routes that exist
param cost {LINKS} >= 0;         # one cost per existing route
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
ampl.set["ORIG"] = ["GARY", "PITT"]
ampl.set["DEST"] = ["FRA", "DET", "LAN"]
ampl.set["LINKS"] = [("GARY", "DET"), ("GARY", "LAN"), ("PITT", "FRA"), ("PITT", "DET")]
ampl.param["cost"] = {("GARY", "DET"): 14, ("GARY", "LAN"): 11,
                      ("PITT", "FRA"): 27, ("PITT", "DET"): 16}
```

```python
# run.py
print(ampl.param["cost"].to_dict())
print(ampl.param["cost"]["PITT", "FRA"])
```

### Pure AMPL: data.dat and model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%writefile data.dat
set ORIG := GARY PITT;
set DEST := FRA DET LAN;
set LINKS := (GARY,DET) (GARY,LAN) (PITT,FRA) (PITT,DET);
param cost :=
  GARY DET  14
  GARY LAN  11
  PITT FRA  27
  PITT DET  16 ;
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display cost;
```
