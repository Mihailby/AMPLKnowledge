## Equality Is a Special Case

An equality on one side of `complements` already holds with equality, so the other expression receives no complementarity restriction. This is valid syntax but usually signals that an ordinary equality would state the intent more clearly.

With `x = 3 complements y`, the pair's `slack` depends only on the equality: it is 0 at $x=3$
whatever value `y` takes, and negative when the equality fails:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
var x;  # first variable
var y;  # free: no restriction from the pair

subject to Eq: x = 3 complements y;  # equality side: usually a plain equality
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# run.py
ampl.option["presolve"] = 0
for x, y in ((3, 7), (3, -5), (2, 7)):
    ampl.var["x"].set_value(x)
    ampl.var["y"].set_value(y)
    print(x, y, ampl.get_value("Eq.slack"))
```
