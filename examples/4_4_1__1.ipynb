## Wrapping Around a Cycle

Every member has a next and a previous member; none is first in the sense of having nothing
before it.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set SHIFTS circular;  # work shifts; the last is followed by the first
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["SHIFTS"] = ["morning", "evening", "night"]
```

```python
# run.py
print(ampl.set["SHIFTS"].to_list())
print(ampl.get_value("next('night', SHIFTS)"))
print(ampl.get_value("prev('morning', SHIFTS)"))
```

### Pure AMPL: data.dat and model.run

Pure AMPL:

```python
%%writefile data.dat
set SHIFTS := morning evening night;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display next('night', SHIFTS), prev('morning', SHIFTS);
```
