## Syntax for Sets

| Attribute | Meaning | Page |
|---|---|---|
| `dimen n` | members are tuples of `n` items: `dimen 2` for pairs | Flat vs Indexed Sets (AMPL lab documentation, section 1.1.1) |
| `within sexpr` | every member must also be in the set expression | Flat vs Indexed Sets (AMPL lab documentation, section 1.1.1) |
| `= sexpr` | the set is defined by an expression; data can't change it | Set Declaration vs. Definition (AMPL lab documentation, section 1.1.2) |
| `default sexpr` | members to use when the data gives none | Set Declaration vs. Definition (AMPL lab documentation, section 1.1.2) |
| `ordered`, `circular`, `ordered by sexpr` | the members keep an order | Ordered Sets and Time Periods (AMPL lab documentation, section 4.4) |

The simplest declaration is just a name. The members come later, from data. `ampl.set["PROD"]` (or
`ampl.get_set("PROD")`) is the `Set` object: assigning a list to it gives the set its members,
`to_list()` returns them as a Python list, and `size()` counts them:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD;   # products; members come from data
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PROD"] = ["bands", "coils", "plate"]
```

```python
# run.py
print(ampl.set["PROD"].to_list())
print(ampl.set["PROD"].size())
```

### Pure AMPL: data.dat and model.run

Pure AMPL. The data file lists the members after `:=`, and
`card(PROD)` is the number of members:

```python
%%writefile data.dat
set PROD := bands coils plate;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display PROD, card(PROD);
```
