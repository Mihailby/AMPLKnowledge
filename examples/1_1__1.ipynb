### AMPL Formulation
#### Model Fragment

```python
%%ampl_eval     # Run AMPL code cell
reset;          # Clear model, data, and options

set PROD;   # Declare empty set PROD in AMPL
```

#### Data

```python
%%ampl_eval     # Run AMPL code cell
reset data;     # Clear only data, keep model declarations
data;           # Start data mode

set PROD := bands, coils, plate;  # Populate with 3 products (:= not =)

display PROD, card(PROD);         # Show set members and cardinality
```

### AMPLPY Formulation

```python
# Define AMPL set PROD and assign values from a Python list
ampl.reset()            # Clears all AMPL model components, data, and options
ampl.eval("set PROD;")  # Declare empty set PROD in AMPL
ampl.set["PROD"].set_values(["bands", "coils", "plate"])  # Populate with 3 products

members = list(ampl.set["PROD"].members()) # Retrieve set members as Python list
n = ampl.set["PROD"].size()                # Get cardinality (number of elements)

# Display set data, members list, and size
print(f"Set values: {ampl.set['PROD'].getValues()}\nMembers: {members}\nSize: {n}")
```
