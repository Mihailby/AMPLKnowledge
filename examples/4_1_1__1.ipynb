## Checking Members and Subsets

The declaration `set PRIORITY within PRODUCTS;` makes the subset relationship part of the data
contract. `ampl.get_value()` evaluates the logical and cardinality expressions.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PRODUCTS;
set PRIORITY within PRODUCTS;
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PRODUCTS"] = ["bands", "coils", "plate"]
ampl.set["PRIORITY"] = ["bands", "plate"]
```

```python
# run.py
print(ampl.get_value("if 'coils' in PRODUCTS then 1 else 0"))
print(ampl.get_value("if PRIORITY within PRODUCTS then 1 else 0"))
print(ampl.get_value("card(PRIORITY)"))
```

### Pure AMPL: data.dat and model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%writefile data.dat
set PRODUCTS := bands coils plate;
set PRIORITY := bands plate;
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display if 'coils' in PRODUCTS then 1 else 0;
display if PRIORITY within PRODUCTS then 1 else 0;
display card(PRIORITY);
```
