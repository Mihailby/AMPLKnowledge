## Recursive Definitions

An indexed parameter may be defined in terms of other members of the same collection, as long
as each value only needs values computed before it. The number of ways to choose `k` items out
of `n`, read one value at a time with a subscript, `ampl.param["comb"][5, 2]`:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
param N integer > 0;
param comb {n in 0..N, k in 0..n} =
    if k = 0 or k = n then 1 else comb[n-1, k-1] + comb[n-1, k];
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
ampl.param["N"] = 5
```

```python
# run.py
print(ampl.param["comb"][5, 2], ampl.param["comb"][4, 2])
```

### Pure AMPL: data.dat and model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%writefile data.dat
param N := 5;
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display comb[5,2], comb[4,2];
```
