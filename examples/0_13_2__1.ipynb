## Exercise 5-1: Rounding in Numeric Sets

Display the sets `-5/3 .. 5/3 by 1/3` and `0 .. 1 by .1`, and look for evidence of rounding error.
A set `a .. b by c` holds $a, a + c, a + 2c, \ldots$ up to $b$
(Sets of Numbers and Ranges (AMPL lab documentation, section 4.2)). Neither 1/3 nor 0.1 is exact in binary floating point,
so AMPL computes the members with rounding. The model declares both sets; in the run fragments,
`ampl` has already read it.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set THIRDS = -5/3 .. 5/3 by 1/3;   # eleven members from -5/3 to 5/3
set TENTHS = 0 .. 1 by .1;         # eleven members from 0 to 1
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# run.py
print(ampl.get_value("card(THIRDS)"), ampl.get_value("card(TENTHS)"))
print(ampl.set["TENTHS"].to_list())
ampl.eval('printf {t in TENTHS} "%.17g\\n", t;')
```

### Pure AMPL: model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%ampl_eval
reset;
reset options;
model model.mod;
```

```python
%%ampl_eval
# model.run
display card(THIRDS), card(TENTHS);
display TENTHS;
printf {t in TENTHS} "%.17g\n", t;
```
