## Initial Values

`:= expr` gives a variable a starting value set in the model; `default expr` gives one that
the data may override. Either way it's only a starting point: solving replaces it.

In `run.py`, `value()` before the solve shows the initial value. After the solve it shows the
solver's value:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
var x >= 0, <= 10 := 3;   # starts at 3; the solver may move it
maximize f: x;            # push x to its upper bound
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# run.py
x = ampl.var["x"]
print(x.value())              # before solving: the initial value
ampl.solve(solver="highs")
print(x.value())              # after solving: the solver's value
```

### Pure AMPL: model.run

Pure AMPL:

```python
%%ampl_eval
reset;
reset options;
model model.mod;
```

```python
%%ampl_eval
# model.run
display x;            # before solving: the initial value
option solver highs;
solve;
display x;            # after solving: the solver's value
```
