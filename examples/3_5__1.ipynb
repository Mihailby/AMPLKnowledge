## Choose the smallest useful shape

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD;
param profit {PROD};
param market {PROD} >= 0;
var Make {p in PROD} >= 0, <= market[p];
maximize Total_Profit: sum {p in PROD} profit[p] * Make[p];
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PROD"] = ["bands", "coils"]
ampl.param["profit"] = {"bands": 25, "coils": 30}
ampl.param["market"] = {"bands": 6000, "coils": 4000}
```

```python
# run.py
ampl.solve(solver="highs")
if ampl.solve_result != "solved":
    raise RuntimeError(ampl.get_value("solve_message"))

print(ampl.obj["Total_Profit"].value())
print(ampl.var["Make"].to_dict())
print(ampl.get_data("{p in PROD} profit[p] * Make[p]").to_dict())
```

### Pure AMPL: data.dat and model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%writefile data.dat
set PROD := bands coils;
param: profit market :=
  bands 25 6000
  coils 30 4000;
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
option solver highs;
solve;
if solve_result = "solved" then {
  display Total_Profit, Make;
  display {p in PROD} profit[p] * Make[p];
} else print solve_message;
```
