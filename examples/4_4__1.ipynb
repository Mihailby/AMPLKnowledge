## Positions in an Ordered Set

`WEEKS` takes its order from the data. A Python list given to `ampl.set["WEEKS"]` keeps its
order, `to_list()` returns the members in that order, and `ampl.get_value(...)` evaluates one
AMPL expression, here the functions that find a member by its position and a position by its
member:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set WEEKS ordered;  # weeks, in the order given in the data
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["WEEKS"] = ["27sep", "04oct", "11oct", "18oct"]
```

```python
# run.py
print(ampl.set["WEEKS"].to_list())
print(ampl.get_value("first(WEEKS)"), ampl.get_value("last(WEEKS)"))
print(ampl.get_value("member(2, WEEKS)"), ampl.get_value("ord('11oct', WEEKS)"))
print(ampl.get_value("ord0('01nov', WEEKS)"))
```

### Pure AMPL: data.dat and model.run

Pure AMPL. The data file lists the members in order,
and `display` prints the values:

```python
%%writefile data.dat
set WEEKS := 27sep 04oct 11oct 18oct;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display first(WEEKS), last(WEEKS), member(2, WEEKS);
display ord('11oct', WEEKS), ord0('01nov', WEEKS);
```
