## List the Loaded Handlers

AMPL lists the handlers loaded in the current session in the set `_HANDLERS` and their
libraries in `_handler_lib`. A handler from the plugins module is available only after `load`:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD;
param rate {PROD} > 0;
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# run.py
ampl.eval("load amplcsv.dll; load amplxl.dll; load amplsqlite3.dll; load eodbc.dll;")
ampl.display("_HANDLERS")
ampl.display("_handler_lib")
```

### Pure AMPL: model.run

`ampl.display` is used here because AMPL's display of the two system sets is the point. A
handler named in documentation but absent from `_HANDLERS` is not available to the current AMPL
process. Each plugin also publishes its accepted options:
`print _handler_desc["amplcsv"];` prints the option list that CSV Files (AMPL lab documentation, section 9.5.1) summarizes.

Pure AMPL, with the same `model.mod`:

```python
%%ampl_eval
reset;
reset options;
model model.mod;
```

```python
%%ampl_eval
# model.run
load amplcsv.dll;
load amplxl.dll;
load amplsqlite3.dll;
load eodbc.dll;
display _HANDLERS;
display _handler_lib;
```
