## When a Binary Variable Is Needed

The solve reports an unbounded problem instead of 3. Writing `abs` directly lets the MP solver
interface build the disjunctive form with a binary variable, using the bounds on `x`:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
var x >= -3, <= 2;                # value whose magnitude should be largest
maximize Magnitude: abs(x);       # MP reformulates abs for HiGHS
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# run.py
ampl.solve(solver="highs")
print(ampl.solve_result, "x =", ampl.var["x"].value(), " abs =", ampl.obj["Magnitude"].value())
```
