## A Convex Allocation QP

The model allocates a fixed total among activities while penalizing squared deviation from targets. The penalty coefficients are positive, so the Hessian is positive definite and the objective is strictly convex; with a nonempty linear feasible set, the optimizer is unique.

![Elliptical contours of a convex quadratic objective touching a linear feasible segment at the constrained optimum](/assets/docs/quadratic-contours.svg)

*Elliptical level curves are characteristic of a positive-definite quadratic. The constrained optimum is the first contour that touches the feasible set.*

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set ACT;                 # activities that share the total
param target {ACT};      # desired amount per activity
param weight {ACT} > 0;  # penalty per squared unit of deviation
param total >= 0;        # amount to allocate

var Allocate {ACT} >= 0;  # amount given to each activity

minimize WeightedDeviation:  # strictly convex: all weights positive
    sum {a in ACT} weight[a] * (Allocate[a] - target[a])^2;

subject to TotalAllocation:  # allocate exactly the total
    sum {a in ACT} Allocate[a] = total;
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["ACT"] = ["A", "B", "C"]
ampl.param["target"] = {"A": 4, "B": 3, "C": 2}
ampl.param["weight"] = {"A": 1, "B": 2, "C": 4}
ampl.param["total"] = 12
```

```python
# run.py
ampl.solve(solver="highs")
print(ampl.solve_result)
print(ampl.var["Allocate"].to_dict())
print(round(ampl.obj["WeightedDeviation"].value(), 6))
```

### Pure AMPL: data.dat and model.run

Pure AMPL:

```python
%%writefile data.dat
set ACT := A B C;
param target := A 4 B 3 C 2;
param weight := A 1 B 2 C 4;
param total := 12;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
option solver highs;
solve;
display Allocate, WeightedDeviation;
```

### Another way in amplpy

Check the result before using it: confirm `ampl.solve_result` is `solved`, then recompute the
quadratic objective and the linear constraint from the returned values instead of trusting the
reported objective alone. With the same `model.mod` and `data.py`, `to_dict()` gives the values
to recompute from:

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["ACT"] = ["A", "B", "C"]
ampl.param["target"] = {"A": 4, "B": 3, "C": 2}
ampl.param["weight"] = {"A": 1, "B": 2, "C": 4}
ampl.param["total"] = 12
```

```python
# run.py
ampl.solve(solver="highs", verbose=False)
alloc = ampl.var["Allocate"].to_dict()
target = ampl.param["target"].to_dict()
weight = ampl.param["weight"].to_dict()
recomputed = sum(weight[a] * (alloc[a] - target[a]) ** 2 for a in alloc)
print(ampl.solve_result, round(recomputed, 6), round(sum(alloc.values()), 6))
```
