## Supply Data and Solve

In the `model.mod` tab below, AMPL declarations separate the reusable relationships from this
instance's numbers. The set and parameters are data, `Make` is the decision, `TotalProfit` is
the objective, and `TimeLimit` restricts the decision. The variable declaration also gives each
product its own lower and upper bound.

Lists and dictionaries supply the instance; `ampl.solve(solver="highs")` selects
HiGHS. Always check `ampl.solve_result` before reading a solution.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD;                          # Products that can be manufactured

param profit {PROD};               # Profit per ton
param rate {PROD} > 0;             # Production rate in tons per hour
param market {PROD} >= 0;          # Maximum sales in tons
param avail >= 0;                  # Available production hours

var Make {p in PROD} >= 0, <= market[p];  # Tons to manufacture

maximize TotalProfit:              # Total profit in dollars
    sum {p in PROD} profit[p] * Make[p];

subject to TimeLimit:              # Hours used cannot exceed available hours
    sum {p in PROD} Make[p] / rate[p] <= avail;
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PROD"] = ["bands", "coils"]  # Product names
ampl.param["profit"] = {"bands": 25, "coils": 30}  # Dollars per ton
ampl.param["rate"] = {"bands": 200, "coils": 140}  # Tons per hour
ampl.param["market"] = {"bands": 6000, "coils": 4000}  # Tons
ampl.param["avail"] = 40  # Production hours
```

```python
# run.py
ampl.solve(solver="highs")  # Solve the linear program with HiGHS
assert ampl.solve_result == "solved", ampl.solve_result  # Verify solve status
print(ampl.solve_result)  # Report status with the solution
print(ampl.obj["TotalProfit"].value())  # Read the optimal profit
print(ampl.var["Make"].to_pandas())  # Read the production plan
make = ampl.var["Make"].to_dict()  # Detach solution values into a Python dictionary
used_hours = sum(make[p] / ampl.param["rate"][p] for p in make)  # Recompute time use
print(f"used hours: {used_hours:.1f}")  # Compare with the 40-hour limit
print(f"available hours: {ampl.param['avail'].value():.1f}")  # Report the data limit
```

### Pure AMPL: data.dat and model.run

Pure AMPL:

```python
%%writefile data.dat
set PROD := bands coils;

param: profit rate market :=
    bands  25  200  6000
    coils  30  140  4000;

param avail := 40;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
option solver highs;
solve;

display solve_result;
display TotalProfit;
display Make;
display sum {p in PROD} Make[p] / rate[p];
display avail;
```
