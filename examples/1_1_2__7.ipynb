## Quoting Strings

From Python, members are Python strings, so they may hold any characters and need no AMPL
quoting:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set CODES;
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
ampl.set["CODES"] = ["New_York", "27sep", "A&P", "Gary, Indiana"]
```

```python
# run.py
print(ampl.set["CODES"].to_list())
```

### Pure AMPL: data.dat and model.run

In a **data file**, quotes are optional for members made of letters, digits, `_`, `.`, `+` and `-`
that don't read as a number. Anything else needs quotes. The same in pure AMPL, with the same
`model.mod`:

```python
%%writefile data.dat
set CODES := New_York 27sep "A&P" "Gary, Indiana";
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display CODES;
```
