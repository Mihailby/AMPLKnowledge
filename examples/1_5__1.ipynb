## Declaring an Objective

The production model maximizes the total profit of the steel it makes. After
`ampl.solve(solver="highs")`, `ampl.solve_result` tells whether the solve succeeded, and
`ampl.obj["Total_Profit"]` (or `ampl.get_objective("Total_Profit")`) is the `Objective` object:
`value()` returns its value. `ampl.get_data(...)` evaluates any AMPL expression, here the
percentage of the profit that comes from each product, and `to_dict()` turns it into a
dictionary:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD;                    # products
param rate {PROD} > 0;       # tons produced per hour
param profit {PROD};         # profit per ton
param market {PROD} >= 0;    # most that can be sold, in tons
param avail >= 0;            # hours available

var Make {p in PROD} >= 0, <= market[p];   # tons to make

maximize Total_Profit: sum {p in PROD} profit[p] * Make[p];   # total profit

subject to Time: sum {p in PROD} (1 / rate[p]) * Make[p] <= avail;   # hours used <= available
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PROD"] = ["bands", "coils"]
ampl.param["avail"] = 40
ampl.param["rate"] = {"bands": 200, "coils": 140}
ampl.param["profit"] = {"bands": 25, "coils": 30}
ampl.param["market"] = {"bands": 6000, "coils": 4000}
```

```python
# run.py
ampl.solve(solver="highs")
print(ampl.solve_result)
print(ampl.obj["Total_Profit"].value())
share = ampl.get_data("{p in PROD} 100 * profit[p] * Make[p] / Total_Profit")
print(share.to_dict())
```

### Pure AMPL: data.dat and model.run

Pure AMPL. `display` prints the objective's value, and an
indexed expression as a table:

```python
%%writefile data.dat
set PROD := bands coils;
param avail := 40;
param:  rate  profit  market :=
  bands  200    25     6000
  coils  140    30     4000 ;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
option solver highs;
solve;
display Total_Profit;
display {p in PROD} 100 * profit[p] * Make[p] / Total_Profit;
```
