## The Member Before and After

The dashed arrow exists only for `nextw` and `prevw`; `next` of the last member and `prev` of the
first are errors.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set WEEKS ordered;  # weeks, in the order given in the data
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["WEEKS"] = ["27sep", "04oct", "11oct", "18oct"]
```

```python
# run.py
print(ampl.get_value("next('04oct', WEEKS)"), ampl.get_value("prev('04oct', WEEKS)"))
print(ampl.get_value("next('04oct', WEEKS, 2)"))
print(ampl.get_value("nextw('18oct', WEEKS)"), ampl.get_value("prevw('27sep', WEEKS)"))
```

### Pure AMPL: data.dat and model.run

Pure AMPL:

```python
%%writefile data.dat
set WEEKS := 27sep 04oct 11oct 18oct;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display next('04oct', WEEKS), prev('04oct', WEEKS), next('04oct', WEEKS, 2);
display nextw('18oct', WEEKS), prevw('27sep', WEEKS);
```
