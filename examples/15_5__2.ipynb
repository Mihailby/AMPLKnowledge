## Debug an Error with the Server

The call is `search_kb(query="avail has an = assignment in the model", mode="debug")`. Whatever
the assistant answers, check it by running its correction. The AMPL book's rule is that `=` in a
declaration defines the value in the model, so data cannot replace it; the fix is a restriction
such as `>= 0`, and the corrected input runs:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD;               # steel products
param rate {PROD} > 0;  # tons per hour
param avail >= 0;       # rolling hours, set from data
var Make {PROD} >= 0;   # tons produced

maximize Total_Tons: sum {p in PROD} Make[p];  # total output, tons

# rolling hours used fit within avail
subject to Time: sum {p in PROD} (1 / rate[p]) * Make[p] <= avail;
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PROD"] = ["bands", "coils"]
ampl.param["rate"] = {"bands": 200, "coils": 140}
ampl.param["avail"] = 35
```

```python
# run.py
ampl.solve(solver="highs", verbose=False)
print(ampl.solve_result, ampl.obj["Total_Tons"].value())
```
