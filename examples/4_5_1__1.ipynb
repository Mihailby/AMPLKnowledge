## Cartesian Products and Sparse Tuples

`ALL_ROUTES` is the full product. `ROUTES` is the sparse subset supplied by the data.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set ORIGINS;
set DESTINATIONS;
set PRODUCTS;

set ALL_ROUTES = ORIGINS cross DESTINATIONS cross PRODUCTS;
set ROUTES within ALL_ROUTES;
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
ampl.set["ORIGINS"] = ["GARY", "CLEV"]
ampl.set["DESTINATIONS"] = ["DET", "LAN"]
ampl.set["PRODUCTS"] = ["bands", "coils"]
ampl.set["ROUTES"] = [("GARY", "DET", "bands"), ("CLEV", "LAN", "coils")]
```

```python
# run.py
print(ampl.set["ALL_ROUTES"].size())
print(ampl.set["ROUTES"].to_list())
```

### Pure AMPL: data.dat and model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%writefile data.dat
set ORIGINS := GARY CLEV;
set DESTINATIONS := DET LAN;
set PRODUCTS := bands coils;
set ROUTES := (GARY,DET,bands) (CLEV,LAN,coils);
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display card(ALL_ROUTES), ROUTES;
```
