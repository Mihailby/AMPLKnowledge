## Using the AMPL GSL library

AMPL's GSL binding packages hundreds of scientific, statistical, and probability functions in `amplgsl.dll`. The distribution also provides `gsl.ampl`, whose declarations make all packaged functions available.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
load amplgsl.dll;
function gsl_cdf_ugaussian_P;

param z;
param probability = gsl_cdf_ugaussian_P(z);
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
ampl.param["z"] = 0
```

```python
# run.py
print(ampl.param["probability"].value())
```

### Pure AMPL: data.dat and model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%writefile data.dat
param z := 0;
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display probability;
```
