## A Two-Dimensional Parameter

In Python, a dictionary with tuple keys states the index association explicitly.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set ORIG;
set DEST;
param cost {ORIG, DEST} >= 0;
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
ampl.set["ORIG"] = ["GARY", "CLEV"]
ampl.set["DEST"] = ["FRA", "DET"]
ampl.param["cost"] = {
    ("GARY", "FRA"): 39,
    ("GARY", "DET"): 14,
    ("CLEV", "FRA"): 27,
    ("CLEV", "DET"): 9,
}
```

```python
# run.py
print(ampl.param["cost"].to_dict())
```

### Pure AMPL: data.dat and model.run

The same values in table form. Each cell is `cost[row_label,column_header]`:

```python
%%writefile data.dat
set ORIG := GARY CLEV;
set DEST := FRA DET;
param cost: FRA DET :=
  GARY 39 14
  CLEV 27  9;
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display cost;
```
