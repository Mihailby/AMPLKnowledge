## Precision is not accuracy

In `data.py` and `run.py`, `ampl` is an `AMPL` object that has already read `model.mod`. From
Python, set AMPL options through `ampl.option[...]`. `ampl.get_output()` returns what a
`display` prints, while `value()` returns the stored number:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
param third = 1 / 3;
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
# This model has no external data.
```

```python
# run.py
ampl.option["display_round"] = 4
print(ampl.get_output("display third;"), end="")
print(ampl.param["third"].value())
print(ampl.get_value("3 * third"))
```

### Pure AMPL: data.dat and model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%writefile data.dat
# This model has no external data.
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
option display_round 4;
display third;
option display_round '';
display third, 3 * third;
```
