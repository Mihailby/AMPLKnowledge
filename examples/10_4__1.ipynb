## A two-piece overtime cost

The first 40 hours cost 20 per hour and later hours cost 35 per hour:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
param required_hours >= 0;  # hours of work needed
var Hours >= required_hours;  # hours worked
minimize LaborCost: <<40; 20, 35>> Hours;  # 20 per hour up to 40 hours, 35 beyond
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.param["required_hours"] = 48
```

```python
# run.py
ampl.solve(solver="highs")
print(ampl.solve_result)
print(ampl.var["Hours"].value())
print(ampl.obj["LaborCost"].value())
```

### Pure AMPL: data.dat and model.run

The same in pure AMPL:

```python
%%writefile data.dat
param required_hours := 48;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
option solver highs;
solve;
display Hours, LaborCost;
```
