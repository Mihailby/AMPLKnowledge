## Stop at a Threshold

This loop raises a trial value until it reaches a limit, printing only the even values.
`ampl.eval()` is used because a compound AMPL loop has no entity method.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
param trial integer default 0;      # current trial value, raised by the loop
param limit integer > 0 default 6;  # value at which the loop stops
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# run.py
ampl.eval("""
repeat sweep {
    let trial := trial + 1;
    if trial mod 2 = 1 then continue sweep;
    printf "trial %d\\n", trial;
    if trial >= limit then break sweep;
};
""")
```

### Pure AMPL: model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%ampl_eval
reset;
reset options;
model model.mod;
```

```python
%%ampl_eval
# model.run
repeat sweep {
    let trial := trial + 1;
    if trial mod 2 = 1 then continue sweep;
    printf "trial %d\n", trial;
    if trial >= limit then break sweep;
};
```
