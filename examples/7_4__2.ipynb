## When a Binary Variable Is Needed

Enforcing equality on the nonconvex side, for example maximizing `abs(x)` or requiring
`t = min(e1,e2)` without maximizing `t`, needs a disjunction, finite bounds and binary variables.
The epigraph rows give the wrong answer. Here the largest `abs(x)` for `-3 <= x <= 2` is 3, but
maximizing `t` over the epigraph lets `t` grow without limit:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
var x >= -3, <= 2;                # value whose magnitude should be largest
var t;                            # meant to be abs(x)
subject to AbsPos: t >= x;
subject to AbsNeg: t >= -x;
maximize Magnitude: t;            # pushes t up, away from abs(x)
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# run.py
ampl.solve(solver="highs")
print(ampl.solve_result)
```
