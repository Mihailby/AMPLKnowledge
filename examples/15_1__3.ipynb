## Diagnose in layers

Return that complete message to the assistant. AMPL writes the indexing of `sum` in braces, and
the corrected model reads and defines the objective:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD;              # steel products
param profit {PROD};   # dollars per ton
var Make {PROD} >= 0;  # tons produced

maximize Total_Profit: sum {p in PROD} profit[p] * Make[p];  # indexing of sum in braces
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# run.py
print([name for name, obj in ampl.get_objectives()])
```
