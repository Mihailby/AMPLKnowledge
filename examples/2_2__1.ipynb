## One Set and One Parameter

Python lists and dictionaries express the same key-value structure directly. `to_dict()` makes
the mapping visible after it is loaded.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD;             # Products listed in the data
param profit {PROD};  # Profit per ton
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PROD"] = ["bands", "coils", "plate"]
ampl.param["profit"] = {"bands": 25, "coils": 30, "plate": 29}
```

```python
# run.py
print(ampl.param["profit"].to_dict())
```

### Pure AMPL: data.dat and model.run

The same data as lists in `data.dat`:

```python
%%writefile data.dat
set PROD := bands coils plate;
param profit :=
  bands 25
  coils 30
  plate 29;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display profit;
```
