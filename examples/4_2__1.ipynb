## Parameterized Ranges

Use data parameters rather than fixed literals when the planning horizon changes between runs.
Because `YEARS` contains numbers, arithmetic such as `y - step` names the previous planning year.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
param first_year integer;               # first planning year
param last_year >= first_year integer;  # last planning year, not before the first
param step > 0 integer;                 # years between planning points

set YEARS = first_year .. last_year by step;  # planning years, every step years
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.param["first_year"] = 2025
ampl.param["last_year"] = 2043
ampl.param["step"] = 5
```

```python
# run.py
print(ampl.set["YEARS"].to_list())
```

### Pure AMPL: data.dat and model.run

Pure AMPL:

```python
%%writefile data.dat
param first_year := 2025;
param last_year := 2043;
param step := 5;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display YEARS;
```
