## `default`: a Starting Value the Data May Replace

`default sexpr` gives members only when the data gives none. Use it when a set usually has the same
members but some data sets need others. Members assigned from Python replace the default too:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD default {"bands", "coils"};
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PROD"] = ["plate"]
```

```python
# run.py
print(ampl.set["PROD"].to_list())
```

### Pure AMPL: data.dat and model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%writefile data.dat
set PROD := plate;
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display PROD;
```
