## A Single Constraint

In the production model, making a ton of product `p` takes `1/rate[p]` hours, and only
`avail` hours are available. In `data.py` and `run.py`, `ampl` is an `AMPL` object that has
already read `model.mod`. After `ampl.solve(solver="highs")`, `ampl.solve_result` is `"solved"`
when the solver found an optimal solution, and `ampl.con["Time"]` (or
`ampl.get_constraint("Time")`) is the `Constraint` object. Its `body()` is the value of the
constraint's expression (hours used), `slack()` is how far it is from its limit, and `dual()` is
its **dual value**: here, the extra profit one more hour would bring (see
Duals, Reduced Costs and Slacks (AMPL lab documentation, section 3.6)):

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set PROD;
param rate {PROD} > 0;
param profit {PROD};
param market {PROD} >= 0;
param avail >= 0;
var Make {p in PROD} >= 0, <= market[p];

maximize Total_Profit: sum {p in PROD} profit[p] * Make[p];

subject to Time: sum {p in PROD} (1 / rate[p]) * Make[p] <= avail;
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# data.py
ampl.set["PROD"] = ["bands", "coils"]
ampl.param["avail"] = 40
ampl.param["rate"] = {"bands": 200, "coils": 140}
ampl.param["profit"] = {"bands": 25, "coils": 30}
ampl.param["market"] = {"bands": 6000, "coils": 4000}
```

```python
# run.py
ampl.solve(solver="highs")
print(ampl.solve_result)

time = ampl.con["Time"]
print(time.body(), time.slack(), time.dual())
```

### Pure AMPL: data.dat and model.run

The same in pure AMPL, with the same `model.mod`. After `solve`, the suffixes `Time.body` and
`Time.slack` give the hours used and the distance to the limit, and the constraint's name on its
own, `Time`, stands for its dual value:

```python
%%writefile data.dat
set PROD := bands coils;
param avail := 40;
param:  rate  profit  market :=
  bands  200    25     6000
  coils  140    30     4000 ;
```

```python
%%ampl_eval
reset;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
option solver highs;
solve;
display Time.body, Time.slack, Time;
```
