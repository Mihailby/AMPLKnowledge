## Pair Sets as Plus/Minus Tables

A two-dimensional set can use a similar table with `+` for a member and `-` for an absent pair.
The colon is required in this form; `(tr)` swaps the row and column roles. From Python, the same
set is a list of tuples:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
set ORIG;                       # Origins
set DEST;                       # Destinations
set LINKS within {ORIG, DEST};  # Routes that exist, a subset of all pairs
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.set["ORIG"] = ["GARY", "CLEV"]
ampl.set["DEST"] = ["FRA", "DET"]
ampl.set["LINKS"] = [("GARY", "DET"), ("CLEV", "FRA"), ("CLEV", "DET")]
```

```python
# run.py
print(ampl.set["LINKS"].to_list())
```

### Pure AMPL: data.dat and model.run

The same set as a plus/minus table in `data.dat`:

```python
%%writefile data.dat
set ORIG := GARY CLEV;
set DEST := FRA DET;
set LINKS: FRA DET :=
  GARY  -   +
  CLEV  +   +;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display LINKS;
```
