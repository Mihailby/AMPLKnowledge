## Restricting a Variable to Chosen Values

`in sexpr` restricts a variable to a set of numbers, given as discrete values and intervals.
Here a tank can only be bought in a few sizes. In `run.py`, `ampl.var["Size"].value()` reads the
chosen size:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
var Size in {0, 10, 20, 50};      # tank size in cubic meters, 0 = no tank
minimize Cost: 300 * Size;
subject to Need: Size >= 15;
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.read("model.mod")
```

```python
# run.py
ampl.solve(solver="highs")
print(ampl.solve_result)
print(ampl.var["Size"].value())
```

### Pure AMPL: model.run

The same in pure AMPL, with the same `model.mod`:

```python
%%ampl_eval
reset;
model model.mod;
```

```python
%%ampl_eval
# model.run
option solver highs;
solve;
display Size;
```
