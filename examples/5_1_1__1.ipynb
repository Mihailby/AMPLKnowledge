## Rounding a Calculation

This example turns a continuous shipment into pack counts and reports a decimal average. The
data values remain unchanged; each computed parameter applies a different function.

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
param units >= 0;                          # Units to pack
param pack_size > 0;                       # Units per pack
param observed;                            # Measured value
param target;                              # Planned value

param packs_down = floor(units / pack_size);  # Full packs only (round down)
param packs_up = ceil(units / pack_size);  # Packs needed for every unit (round up)
param average_2dp = round(units / pack_size, 2);  # Units per pack, 2 decimals
param absolute_error = abs(observed - target);  # Size of the miss, sign dropped
```

### amplpy: data.py and run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `data.py` sets the data from Python and `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# data.py
ampl.param["units"] = 29
ampl.param["pack_size"] = 12
ampl.param["observed"] = 18.5
ampl.param["target"] = 20
```

```python
# run.py
for name in ["packs_down", "packs_up", "average_2dp", "absolute_error"]:
    print(f"{name}: {ampl.get_value(name):g}")
```

### Pure AMPL: data.dat and model.run

Pure AMPL:

```python
%%writefile data.dat
param units := 29;
param pack_size := 12;
param observed := 18.5;
param target := 20;
```

```python
%%ampl_eval
reset;
reset options;
model model.mod;
data data.dat;
```

```python
%%ampl_eval
# model.run
display packs_down, packs_up, average_2dp, absolute_error;
```
