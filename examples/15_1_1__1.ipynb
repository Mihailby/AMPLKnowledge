The smallest run that proves an environment can execute a generated model has one variable and
one constraint:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
var x >= 0;                # single decision
maximize Value: x;         # push x up to its limit
subject to Limit: x <= 1;  # optimum at x = 1
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# run.py
ampl.solve(solver="highs", verbose=False)
print(ampl.solve_result)
print(ampl.var["x"].value())
```
