## Diagnose an Unbounded Model

Check objective direction, missing bounds, and missing linking constraints. For example, this
maximization has no upper bound on `x`:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
var x >= 0;          # No upper bound
maximize Reward: x;  # Grows without limit: unbounded
```
