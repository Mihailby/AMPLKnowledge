## Solve Nonlinear Feasibility Systems

AMPL can also express a system of nonlinear equations or inequalities with no objective. In that
case the task is to find a feasible point, rather than to compare feasible points by an objective.
Ipopt returns one point on the circle arc with $x\ge y$. The point
$(1/\sqrt2,1/\sqrt2)\approx(0.707107,0.707107)$ also satisfies both constraints, so the returned
point is one of many solutions:

### model.mod
The model, written to a file that both ways of running it read.

```python
%%writefile model.mod
var x >= 0, := 1;  # first coordinate, start 1
var y >= 0, := 1;  # second coordinate, start 1

subject to Circle: x^2 + y^2 = 1;  # stay on the unit circle
subject to Balance: x >= y;        # keep the arc where x >= y
```

### amplpy: run.py

`ampl` is the `AMPL` object created above. Each run starts from a fresh session that has read `model.mod`; then `run.py` solves and reads the results.

```python
# A fresh session that has read model.mod, as data.py and run.py assume
ampl.reset()
ampl.eval("reset options;")  # reset keeps the options
ampl.read("model.mod")
```

```python
# run.py
ampl.solve(solver="ipopt", verbose=False)
print(ampl.solve_result)
print(round(ampl.var["x"].value(), 6), round(ampl.var["y"].value(), 6))
print(round(ampl.con["Circle"].body(), 6))
```
